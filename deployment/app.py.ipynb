```python
import streamlit as st
import joblib

# Page configuration
st.set_page_config(page_title="Researchly Classifier", layout="wide")

# App title and description
st.title("Researchly Research Abstract Classifier")
st.markdown("This app classifies research abstracts as either AI-related or classical research.")

# Load the trained model
@st.cache_resource
def load_model():
    return joblib.load('best_inflation_classifier.pkl')

try:
    model = load_model()
except Exception as e:
    st.error(f"Error loading model: {e}")
    st.stop()

# Text area for input
st.markdown("**Enter research abstract:**")
abstract_input = st.text_area(
    label="Abstract Input",
    label_visibility="collapsed",
    height=250,
    placeholder="Paste your research abstract here..."
)

# Button to trigger classification
if st.button("Classify Abstract", type="primary"):
    if abstract_input.strip() == "":
        st.warning("Please enter an abstract to classify.")
    else:
        # Make prediction
        prediction = model.predict([abstract_input])[0]
        
        # Display result
        st.markdown("### Prediction")
        if prediction == 1:
            st.success("The abstract is classified as: **artificial intelligence**")
        else:
            st.info("The abstract is classified as: **classical research**")
```
